### `Descripción de la asignatura`

- La mayoría de las tareas requieren de una **persona o un sistema automático que razone** (que llegue a conclusiones con base en la información disponible).

- El enfoque de **modelos gráficos probabilísticos** constituye un esquema general para dar solución a esta problemática.

- Es un enfoque basado en modelo, de manera que **puedan construirse modelos interpretables** y luego manipularlos por medio de algoritmos de razonamiento (inferencia). Dado que la incertidumbre es un aspecto inevitable de la mayoría de aplicaciones en la vida real, **los modelos que se consideran son probabilísticos**.

- El curso recorre esta idea de principio a fin: parte de los fundamentos probabilísticos y el ajuste de modelos (máxima verosimilitud, enfoque Bayesiano), pasa por su representación gráfica más difundida —las **redes Bayesianas**— junto con sus algoritmos de inferencia y aprendizaje, se extiende hacia la **toma de decisiones bajo incertidumbre** (redes de decisión) y cierra con una familia de modelos con **variables latentes** (mezclas Gaussianas, modelos ocultos de Markov) que comparten el mismo lenguaje gráfico usado para clasificación, agrupamiento (clustering) y datos secuenciales.

- A lo largo del curso trabajamos tanto con ejemplos de juguete —que permiten aislar el concepto que se está enseñando— como con **datasets académicos y reales** (por ejemplo, datos de vivienda, enfermedades cardiacas, calidad de vino, o series de tiempo), usando librerías del ecosistema de Python como PyMC y pgmpy.

- Los modelos gráficos probabilísticos son la base de varios métodos **en varias aplicaciones** tales como diagnóstico médico, aprobación de crédito, mantenimiento predictivo, visión artificial, reconocimiento de voz, procesamiento de lenguaje natural y muchos otros más. Además, son una herramienta básica para la formulación de varios problemas de aprendizaje de máquinas.

### `Objetivos`

- Modelar problemáticas/situaciones reales mediante grafos probabilísticos (redes Bayesianas, redes de decisión y modelos de variable latente).
- Responder preguntas de interés práctico acerca de las problemáticas/situaciones modeladas usando algoritmos de inferencia (exacta y aproximada).
- Estimar los parámetros y/o estructura de los modelos usando técnicas de optimización a partir de datos.
- Tomar decisiones bajo incertidumbre integrando explícitamente creencias (probabilidad) y preferencias (utilidad).
- Reconocer y aplicar la variable latente como principio unificador detrás de tareas aparentemente distintas: clasificación (Naive Bayes), agrupamiento (mezclas Gaussianas) y modelado de datos secuenciales (modelos ocultos de Markov).
- Implementar y ajustar estos modelos sobre datasets académicos y reales usando herramientas del ecosistema de Python (PyMC, pgmpy).

#### Módulo 2. Representación, inferencia, aprendizaje y toma de decisiones en redes Bayesianas.

<img style="float: right; margin: 0px 0px 15px 15px;" src="assets/BayesNets.png" width="500px" height="500px" />

En este módulo aprenderemos acerca de una representación de modelos gráficos probabilísticos: 
las redes Bayesianas (grafos dirigidos). 
Estudiaremos tanto las propiedades teóricas de dichas representaciones así como su uso en la práctica. 
Así mismo, aprenderemos cómo responder preguntas usando las estructuras gráficas y cómo aprender 
sus parámetros a partir de datos.

Cerraremos el módulo extendiendo estas representaciones hacia **redes de decisión**, que nos 
permiten modelar procesos de toma de decisión bajo incertidumbre incorporando explícitamente 
preferencias (utilidad) y no solo creencias (probabilidad).

6. Redes Bayesianas.

   - Factores y operaciones con factores.
   - Redes Bayesianas y modelos probabilísticos.
   - Flujo de influencia probabilística en redes Bayesianas.

7. Inferencia en redes Bayesianas.

   - Introducción a inferencia probabilística.
   - Algoritmo de eliminación de variables.
   - Inferencia aproximada.

8. Aprendizaje en redes Bayesianas.

   - Estimación de parámetros en redes Bayesianas.

9. Redes de decisión.

   - De redes Bayesianas a redes de decisión: nodos de azar, decisión y utilidad.
   - Utilidad esperada y cálculo de la política óptima.
   - Valor de la información.
   - Caso aplicado: una decisión bajo incertidumbre con datos reales (p. ej. aprobación de
     crédito, decisión clínica o mantenimiento predictivo), integrando lo aprendido en los
     capítulos 6-8.

#### Módulo 3. Aplicaciones seleccionadas: modelos de variable latente.

<img style="float: right; margin: 0px 0px 15px 15px;" src="assets/LearningBot.png" width="500px" height="500px" />

En este módulo utilizaremos nuestro conocimiento en redes Bayesianas para estudiar una familia 
de modelos en los que una variable no observada (latente) explica la estructura de los datos 
observados.

Partiremos de un caso discreto y sencillo — el clustering — y avanzaremos hacia 
datos secuenciales. Cerraremos el módulo con una síntesis que conecta explícitamente esta 
familia de modelos con las redes Bayesianas estudiadas en el Módulo 2.

10. Modelos de variables latentes para clustering.

    - K-Medias.
    - Mezclas Gaussianas y algoritmo de maximización de la esperanza (EM).
    - Caso aplicado con datos reales (p. ej. Old Faithful, Palmer Penguins o segmentación
      de clientes).

11. Modelos ocultos de Markov.

    - Variables latentes en datos secuenciales.
    - Algoritmos forward-backward y Viterbi.
    - Caso aplicado con datos reales (p. ej. series de tiempo, regímenes de mercado o clima).

12. Síntesis: de Naive Bayes a modelos de variable latente. (material adicional)

    - Naive Bayes como red Bayesiana con variable de clase.
    - Conexión entre Naive Bayes y mezclas Gaussianas.
    - Cierre del curso: un mismo lenguaje gráfico para clasificación, clustering y series de tiempo.
